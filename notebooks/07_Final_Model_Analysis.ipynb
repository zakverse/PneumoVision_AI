## Mengapa Evaluasi Test Set Dilakukan Paling Akhir?
Data test adalah **data yang belum pernah dilihat sama sekali** oleh model selama proses training maupun tuning hyperparameter. Melakukan evaluasi hanya satu kali di akhir memastikan bahwa metrik performa yang kita peroleh benar-benar mencerminkan kemampuan generalisasi model pada dunia nyata tanpa bias (*data leakage*).

---

## Agenda Utama Notebook Ini
1. **Pemilihan Model Terbaik**: Menentukan model final berdasarkan bukti metrik validasi yang konsisten.
2. **Evaluasi Akhir Test Set**: Menghitung metrik final (Accuracy, Precision, Recall, F1, ROC-AUC) dan menyimpannya ke `results/metrics/`.
3. **Implementasi Grad-CAM**: Mengambil gradien layer konvolusi terakhir untuk memetakan heatmap area paru-paru yang dicurigai.
4. **Kurasi Aset Demo Mode**: Memilih 4 sampel rontgen representatif (2 NORMAL dan 2 PNEUMONIA) beserta JSON hasil inferensinya untuk disimpan ke folder `demo/`.
5. **Ekspor Bobot Model**: Menyimpan bobot produksi `final_pneumovision_model.pth` dan konfigurasi JSON untuk digunakan oleh FastAPI backend.
