## Mengapa Hyperparameter Perlu Disesuaikan?
Arsitektur model yang sama bisa memberikan hasil yang jauh berbeda tergantung konfigurasi latihnya. Nilai Learning Rate yang terlalu besar dapat membuat loss meledak (*diverge*), sedangkan nilai yang terlalu kecil membuat pelatihan sangat lambat dan rawan terjebak di local minima.

---

## Parameter Kunci yang Diuji
1. **Learning Rate & LR Scheduler**: Menguji penggunaan Cosine Annealing atau ReduceLROnPlateau agar nilai learning rate mengecil seiring berjalannya epoch.
2. **Optimizer**: Membandingkan performa Adam, AdamW, dan SGD dengan momentum.
3. **Weight Decay & Dropout**: Menambahkan regularisasi L2 dan dropout pada classification head untuk meminimalkan Overfitting.
4. **Ukuran Batch (Batch Size)**: Menemukan titik keseimbangan kestabilan gradien dan konsumsi memori GPU.
5. **Early Stopping & Checkpoint**: Menyimpan model terbaik secara otomatis berdasarkan validasi metric (misal: validation loss terendah atau F1-score tertinggi).
