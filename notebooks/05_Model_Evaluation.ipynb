## Mengapa Metrik Accuracy Saja Tidak Cukup?
Dalam kasus medis, jenis kesalahan memiliki bobot risiko yang sangat berbeda:
- **False Negative (Tipe II)**: Pasien sebenarnya menderita pneumonia, tetapi diprediksi Normal oleh model. Ini **sangat berbahaya** karena pasien bisa terlambat ditangani.
- **False Positive (Tipe I)**: Pasien sebenarnya sehat, tetapi diprediksi Pneumonia. Pasien mungkin perlu pemeriksaan lanjutan, namun tidak membahayakan nyawa secara langsung.

Oleh sebab itu, kita harus memprioritaskan metrik **Recall (Sensitivity)** tinggi tanpa mengorbankan **Precision** dan **Specificity** secara berlebihan.

## Agenda Utama Notebook Ini

4. **Tabel Komparasi**: Menyusun ringkasan perbandingan performa antar model untuk menentukan kandidat model terbaik.
